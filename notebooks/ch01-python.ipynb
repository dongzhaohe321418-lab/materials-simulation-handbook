## 1.1 NumPy basics — vectorisation beats loops


```python
import numpy as np
import time

N = 1_000_000
x = np.arange(N, dtype=float)

t0 = time.perf_counter()
total = 0.0
for v in x[:50_000]:
    total += v * v
loop_t = time.perf_counter() - t0

t0 = time.perf_counter()
_ = (x * x).sum()
vec_t = time.perf_counter() - t0

print(f'loop (50k items):  {loop_t * 1000:.1f} ms')
print(f'vectorised ({N}): {vec_t * 1000:.2f} ms')
```

## 1.2 Broadcasting — pairwise distance matrix without a loop


```python
import numpy as np

rng = np.random.default_rng(0)
pts = rng.uniform(0, 1, size=(6, 3))
diff = pts[:, None, :] - pts[None, :, :]
dist = np.linalg.norm(diff, axis=-1)
print(np.round(dist, 3))
```

## 1.3 Matplotlib — a publication-grade figure


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 400)
fig, ax = plt.subplots(figsize=(6, 4))
for k in (1, 2, 3):
    ax.plot(x, np.sin(k * x) / k, label=f'sin({k}x)/{k}')
ax.set_xlabel('x')
ax.set_ylabel('amplitude')
ax.set_title('A small family of sinusoids')
ax.legend(frameon=False)
ax.grid(alpha=0.3)
fig.tight_layout()
plt.show()
```

## 1.4 An ipywidgets slider — interactive learning

Move the slider and re-run the cell to change the frequency. If the slider does not appear, run the cell once more after the kernel finishes loading.

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

x = np.linspace(0, 2 * np.pi, 400)

@interact(k=FloatSlider(min=0.5, max=5.0, step=0.1, value=1.0))
def plot_sine(k):
    fig, ax = plt.subplots(figsize=(6, 3))
    ax.plot(x, np.sin(k * x))
    ax.set_ylim(-1.1, 1.1)
    ax.set_title(f'sin({k:.1f} x)')
    plt.show()
```
