## 0.1 Linear algebra — eigenvalues of a symmetric matrix

The single most important operation in computational materials science. Every Hamiltonian diagonalisation, every principal-component analysis, every spectral graph technique reduces to this.

```python
import numpy as np

rng = np.random.default_rng(0)
A = rng.standard_normal((5, 5))
A = (A + A.T) / 2  # symmetrise
eigvals, eigvecs = np.linalg.eigh(A)
print('eigenvalues:', np.round(eigvals, 4))
print('orthonormality check:', np.allclose(eigvecs.T @ eigvecs, np.eye(5)))
```

## 0.2 Calculus — gradient descent on a paraboloid

The workhorse of geometry optimisation and machine-learning training. Tweak the learning rate and watch convergence change.

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return 0.5 * (x[0] ** 2 + 10.0 * x[1] ** 2)

def grad(x):
    return np.array([x[0], 10.0 * x[1]])

x = np.array([3.0, 1.5])
lr = 0.1
trail = [x.copy()]
for _ in range(40):
    x = x - lr * grad(x)
    trail.append(x.copy())
trail = np.array(trail)

xs, ys = np.meshgrid(np.linspace(-3.5, 3.5, 80), np.linspace(-2, 2, 80))
zs = 0.5 * (xs ** 2 + 10.0 * ys ** 2)
fig, ax = plt.subplots(figsize=(6, 4))
ax.contour(xs, ys, zs, levels=20, colors='lightgrey')
ax.plot(trail[:, 0], trail[:, 1], 'o-', color='C3', ms=3)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Gradient descent on f(x,y) = (x^2 + 10y^2)/2')
plt.show()
```

## 0.3 Fourier — a square wave from sines

Reciprocal-space thinking is built on this picture: any well-behaved periodic function is a sum of sinusoids.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 1000)
target = np.sign(np.sin(x))

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(x, target, 'k', lw=1.5, label='target')
approx = np.zeros_like(x)
for k in range(1, 12, 2):
    approx += (4 / (np.pi * k)) * np.sin(k * x)
    if k in (1, 3, 7, 11):
        ax.plot(x, approx, label=f'up to k={k}', lw=1)
ax.legend()
ax.set_xlabel('x')
ax.set_ylabel('partial Fourier sum')
plt.show()
```

## 0.4 Probability — the central limit theorem in action

Sums of independent random variables converge to a Gaussian. This is why noise in everything from thermostats to GP priors is normally distributed.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
fig, axes = plt.subplots(1, 3, figsize=(10, 3), sharey=True)
for ax, n in zip(axes, [1, 5, 50]):
    samples = rng.uniform(-1, 1, size=(20000, n)).mean(axis=1)
    ax.hist(samples, bins=60, density=True, alpha=0.7)
    ax.set_title(f'mean of {n} uniforms')
fig.tight_layout()
plt.show()
```
