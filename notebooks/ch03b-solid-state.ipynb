## 3.5.3 Tight-binding 1D chain

With on-site energy $\varepsilon$ and nearest-neighbour hopping $-t$, the band structure is $E(k) = \varepsilon - 2t \cos(ka)$. Diagonalise a finite chain to recover the same shape.

```python
import numpy as np
import matplotlib.pyplot as plt

N = 200          # number of sites (periodic chain)
eps = 0.0
t = 1.0
a = 1.0

H = eps * np.eye(N) - t * (np.eye(N, k=1) + np.eye(N, k=-1))
H[0, -1] = H[-1, 0] = -t  # periodic boundary
energies = np.sort(np.linalg.eigvalsh(H))

k = np.linspace(-np.pi / a, np.pi / a, 400)
E_analytic = eps - 2 * t * np.cos(k * a)

fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(k, E_analytic, 'k-', lw=1.5, label='analytic')
ax.plot(np.linspace(-np.pi, np.pi, N), energies, 'o', ms=3, alpha=0.6, label='finite chain')
ax.set_xlabel('k a')
ax.set_ylabel('E / t')
ax.legend()
ax.set_title(f'Tight-binding band, N = {N}')
plt.show()
```

## 3.5.5 Phonons of a 1D monatomic chain

Spring constant $K$, mass $m$. Dispersion: $\omega(k) = 2\sqrt{K/m}\,|\sin(ka/2)|$.

```python
import numpy as np
import matplotlib.pyplot as plt

K = 1.0
m = 1.0
a = 1.0
k = np.linspace(-np.pi / a, np.pi / a, 400)
omega = 2 * np.sqrt(K / m) * np.abs(np.sin(k * a / 2))

fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(k * a, omega)
ax.set_xlabel('k a')
ax.set_ylabel('omega (sqrt(K/m))')
ax.set_title('Monatomic phonon chain')
ax.grid(alpha=0.3)
plt.show()
```

Compare with the long-wavelength sound speed $c_s = a\sqrt{K/m}$ (slope of the curve at $k = 0$).
